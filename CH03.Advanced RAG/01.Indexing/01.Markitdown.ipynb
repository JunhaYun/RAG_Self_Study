##### Markitdown 이란?

- 어떤 형태 문서 입력하더라도 마크다운으로 변환해주는 라이브러리
- DL/ML 활용이 아닌 Rule-based 임
- 성능은 DL/ML 대비 낮지만 단순하고 빠르다는 장점이 있음
- 마크다운은 토큰 효율성이 좋은편

##### 01.PDF를 마크다운으로 변환하기

1.2.3 형태나 -  형태만 추가되도 마크다운으로 잘 변환이 안되는군. 텍스트가 주로 이루어질때 사용해야겠다.

```python
from markitdown import MarkItDown

source = "data/Portfolio_26.pdf" 

md = MarkItDown()
result = md.convert(source)
print(result.text_content[:1500])
```

```python
from IPython.display import Markdown

display(Markdown(result.text_content[:1500]))
```

##### 02. DOCX 워드 파일을 마크다운으로 변환하기

확실히 더 글이 위주다보니 마크다운으로 변환을 잘하네. 같은 내용인데.

```python
source = "data/Portfolio_24.docx"  # document per local path or URL

md = MarkItDown() # Set to True to enable plugins
result = md.convert(source)
print(result.text_content[:1000])
```

##### 03. 이미지에서 마크다운 형태의 주석 생성하기

```python
from dotenv import load_dotenv
load_dotenv()

from markitdown import MarkItDown
from openai import OpenAI

client = OpenAI()
md = MarkItDown(llm_client=client, llm_model="gpt-4.1")
result = md.convert("data/Schedule.png")
print(result.text_content)
```
